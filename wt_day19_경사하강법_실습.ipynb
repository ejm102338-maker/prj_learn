```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

X = np.random.rand(100)
y = 0.2*X
```

```python
plt.figure(figsize=(6,4))
plt.scatter(X,y)
```

```python
w = np.random.uniform(-1,1)
b = np.random.uniform(-1,1)

pred = w*X+b
```

```python
def plot_pred(pred,y):
    plt.figure(figsize=(6,4))
    plt.scatter(X,y)
    plt.scatter(X,pred)
```

```python
plot_pred(pred,y)
```

```python
lr = 0.5

for epoch in range(300):
    # 기울기 초기화
    w_grad = 0
    b_grad = 0

    # 예측하기
    y_pred = w*X + b

    # 손실 구하기
    loss = np.abs(y_pred - y).mean()

    # 가중치 재 설정
    w_grad = lr*((y_pred-y)*X).mean()
    b_grad = lr*(y_pred-y).mean()

    # 기존 가중치 설정
    w = w - w_grad
    b = b - b_grad

    if loss < 0.001 : 
        break

    print("Loss : ",loss)

    if epoch % 10 == 0:
        plot_pred(y_pred,y)
```
